# 练习1：NumPy array数组与其形状

```python
import numpy as np

# 创建名为 daily_minutes 的 NumPy 数组
daily_minutes = np.array([30, 45, 60, 20, 50, 40, 35])
print(f'本周每天学习时长：{daily_minutes} 分钟')

# 假设每天额外增加 15 分钟，计算新的学习时长
increased_minutes = daily_minutes + 15
print(f'每天额外增加 15 分钟后的学习时长：{increased_minutes} 分钟')

# 将增加后的学习时长转换为小时
daily_hours = increased_minutes / 60
print(f'每天学习时长（小时）：{daily_hours} 小时')

# 一维数组的 shape
print(f'一维数组的形状：{daily_minutes.shape}')

# 二维数组的 shape
weekly_minutes = np.array(
    [[30, 45, 60, 20, 50, 40, 35],
     [25, 30, 55, 15, 45, 35, 30],
     [35, 40, 65, 25, 55, 45, 40]]
)
print(f'二维数组的形状：{weekly_minutes.shape}')
```

# 练习2：NumPy 数组索引

```python
import numpy as np

daily_minutes = np.array([30, 45, 60, 20, 50, 40, 35])

# 负数索引
# 数据：      [60, 75, 90, 45]
# 正数索引：    0   1   2   3
# 负数索引：   -4  -3  -2  -1

# 获取索引为 0 的元素，也就是星期一学习时长
first_day_minutes = daily_minutes[0]
print(f'星期一的学习时长：{first_day_minutes} 分钟')

# 获取索引为 2 的元素，也就是星期三学习时长
third_day_minutes = daily_minutes[2]
print(f'星期三的学习时长：{third_day_minutes} 分钟')

# 获取最后一个元素，也就是星期天学习时长
last_day_minutes = daily_minutes[-1]
print(f'星期天的学习时长：{last_day_minutes} 分钟')

# 获取倒数第三个元素，也就是星期五学习时长
third_last_day_minutes = daily_minutes[-3]
print(f'星期五的学习时长：{third_last_day_minutes} 分钟')

print('=' * 100)

# 二维数组索引--数组名[行索引, 列索引]
# 行索引 第0列 第1列 第2列 第3列
#   0	  60	75	 90	   45
#   1	  80	60	 120   90
#   2	  45	75	 60	   105

Weekly_minutes = np.array(
    [[60, 75, 90, 45, 63],
     [80, 60, 120, 90, 32],
     [45, 75, 60, 105, 34]]
)

# 获取第 1 周第 3 天的学习时长
print(f'第 1 周第 3 天的学习时长：{Weekly_minutes[0, 2]} 分钟')

# 获取第 2 周第 4 天的学习时长
print(f'第 2 周第 4 天的学习时长：{Weekly_minutes[1, 3]} 分钟')

# 获取第 3 周的全部数据
print(f'第 3 周的全部数据：{Weekly_minutes[2]} 分钟')

# 使用负数索引获取最后一周的全部数据
print(f'最后一周的全部数据：{Weekly_minutes[-1]} 分钟')
```

# 练习3：NumPy 数组切片

```python
# 一维数组切片--数组名[开始索引:结束索引]
import numpy as np

daily_minutes = np.array([60, 75, 90, 45, 80])

# 获取索引 1 到索引 3 之前的数据
# 包含索引 1，但不包含索引 3
selected_minutes = daily_minutes[1:3]
print(f'索引 1 到索引 3 之前的数据：{selected_minutes} 分钟')

# 省略开始位置，从数组开头取到索引 3 之前
first_third_minutes = daily_minutes[:3]
print(f'从数组开头取到索引 3 之前的数据：{first_third_minutes} 分钟')

# 省略结束位置，从索引 2 开始取到数组末尾
last_minutes = daily_minutes[2:]
print(f'从索引 2 开始取到数组末尾的数据：{last_minutes} 分钟')

# 获取整个数组
print(f'整个数组的数据 ：{daily_minutes[:]} 分钟')

print('=' * 100)

# 二维数组切片--数组名[行切片, 列切片]
weekly_minutes = np.array(
    [[60, 75, 90, 45, 63],
     [80, 60, 120, 90, 32],
     [45, 75, 60, 105, 34]]
)

# 获取第 1 周的前 3 天数据
print(f'第一周的前三天数据：{weekly_minutes[0:1, 0:3]} 分钟')

# 获取第 2 周从第 2 天到第 4 天的数据
print(f'第二周的第 2 到第 4 天数据：{weekly_minutes[1:2, 1:4]} 分钟')

# 获取前两周的全部数据
print(f'获取前两周的全部数据：\n{weekly_minutes[0:2, :]} 分钟')

# 获取所有周的第 3 天数据
print(f'所有周第三天的数据：\n{weekly_minutes[:, :3]} 分钟')

# 获取后两周的后两天数据
print(f'后两周的后两天数据：\n{weekly_minutes[1:, 3:]} 分钟')
```

# 练习7：使用 drop_duplicates() 删除重复数据

```python
import pandas as pd

csv_path = 'D:/Ai_Learning/python basic/data/raw/study_log.csv'

csv_df = pd.read_csv(csv_path, encoding='utf-8-sig')

clean_df = csv_df.dropna()

duplicate_flags = clean_df.duplicated()
print(f"重复记录检查结果：\n{duplicate_flags}")

duplicate_count = clean_df.duplicated().sum()
print(f"重复记录总数：\n{duplicate_count}")

# 找出后出现的重复记录
duplicate_records = clean_df[clean_df.duplicated()]
print(f"重复记录：\n{duplicate_records}")

# 找出所有参与重复的记录
# keep=False 表示不保留任何一个作为非重复项
all_records = clean_df[clean_df.duplicated(keep=False)]
print(f"所有参与重复的记录：\n{all_records}")

# 删除完全重复的行
no_duplicate = clean_df.drop_duplicates()
print(f"删除重复记录后的数据：\n{no_duplicate}")

# 根据指定字段判断记录是否重复
no_duplicates = clean_df.drop_duplicates(subset=['date', 'week', 'module', 'duration_minutes'])
print(f"根据日期，周次，模块和学习时长删除重复记录后的数据：\n{no_duplicates}")

# keep 决定重复数据中保留哪一条
# 保留第一条
first_keep = clean_df.drop_duplicates(keep='first')
print(f"保留重复数据中的第一条记录后的数据：\n{first_keep}")
# 保留最后一条
last_keep = clean_df.drop_duplicates(keep='last')
print(f"保留重复数据中的最后一条记录后的数据：\n{last_keep}")

# 删除空值和重复记录后，索引会不连续，可以使用 reset_index() 重新生成连续索引
new_index = no_duplicate.reset_index(drop=True)
print(f"删除空值和重复记录后，重新生成连续索引的数据：\n{new_index}")
```

# 练习8：使用条件筛选清洗异常数据

```python
import pandas as pd

csv_path = 'D:/Ai_Learning/python basic/data/raw/study_log.csv'

csv_df = pd.read_csv(csv_path, encoding='utf-8-sig')

clean_df = csv_df.dropna().drop_duplicates().reset_index(drop='True')
print(f"删除空值和重复记录后的数据：\n{clean_df}")

# 判断每一条学习时长是否大于 0和小于 480 分钟
valid_df = (
    (clean_df['duration_minutes'] > 0) & 
     (clean_df['duration_minutes'] <= 480)
)
print(f"学习时长大于 0 和小于 480 分钟的记录：\n{valid_df}")

# 只保留学习时长大于 0 和小于 480 分钟的记录
positive_df= clean_df[valid_df]
print(f"只保留学习时长大于 0 和小于 480 分钟的记录后的数据：\n{positive_df}")

# 在前面加 ~，表示将判断结果取反
# 找出不符合有效时长规则的记录
invalid_df = clean_df[~valid_df]
print(f"不符合有效时长规则的记录：\n{invalid_df}")

new_index = positive_df.reset_index(drop='True')
print(f"删除无效学习时长记录后，重新生成连续索引的数据：\n{new_index}")
```

# 练习9：使用 groupby() 按周统计学习时长

```python
import pandas as pd

csv_path = 'D:/Ai_Learning/python basic/data/raw/study_log.csv'

csv_df = pd.read_csv(csv_path, encoding='utf-8-sig')

clean_df = csv_df. dropna().drop_duplicates().reset_index(drop='True')
print(f"删除空值和重复记录后的数据：\n{clean_df}")

# groupby() 用于按照某一列中的类别，把数据分成若干组
# 基本写法：DataFrame.groupby("分组列")["统计列"].sum()
weekly_total = clean_df.groupby('week')['duration_minutes'].sum()
print(f"每周学习时长总和：\n{weekly_total}")

# 为了让统计结果更像普通表格，可以重置索引
weekly_index = clean_df.groupby('week')['duration_minutes'].sum().reset_index()
print(f"每周学习时长总和（重置索引后）：\n{weekly_index}")

# 修改统计列名称，当前列名仍然是：duration_minutes。为了表达得更准确，可以将其改为：total_minutes
weekly_column = clean_df.groupby('week')['duration_minutes'].sum().reset_index().rename(columns={'duration_minutes': 'total minutes'})
print(f"每周学习时长统计：\n{weekly_column}")

# 同时计算总时长和平均时长，groupby() 还可以配合 .agg() 进行多项统计
# sum：每周总学习时长，mean：每条记录的平均学习时长
weekly_df = clean_df.groupby('week')['duration_minutes'].agg(['sum', 'mean']).reset_index()
print(f'同时计算总学习时长和平均时长：\n{weekly_df}')
```

# 练习10：使用 value_counts() 统计类别出现次数

```python
import pandas as pd

csv_path = 'D:/Ai_Learning/python basic/data/raw/study_log.csv'

csv_df = pd.read_csv(csv_path, encoding='utf-8-sig')

clean_df = csv_df.dropna().drop_duplicates().reset_index(drop=True)
print(f'删除空值和重复记录后的数据：\n{clean_df}')

# 本节统计 module 列中每个学习模块出现了多少次
# 基本写法：DataFrame["列名"].value_counts()
module_count = clean_df['module'].value_counts(ascending=False) # ascending=False 表示按照次数从高到低排列，True相反
print(f'各模块学习记录次数：\n{module_count}')

# 将统计结果转换为 DataFrame，并且修改统计列名称
module_df = clean_df['module'].value_counts(ascending=True).reset_index().rename(columns={'count': 'record_count'})
print(f'各模块学习记录次数：\n{module_df}')

# 统计各模块所占比例
# normalize=True 表示返回比例，而不是记录数量
module_proportions = clean_df['module'].value_counts(normalize=True) * 100
print(f"各模块记录占比：{module_proportions}")
```

# 练习11：使用 sort_values() 对数据排序

```python
import pandas as pd

csv_path = 'D:/Ai_Learning/python basic/data/raw/study_log.csv'

csv_df = pd.read_csv(csv_path, encoding='utf-8-sig')

clean_df = csv_df.dropna().drop_duplicates().reset_index(drop=True)
print(f'删除空值和重复记录后的数据：\n{clean_df}')

# sort_values() 的基本写法
# DataFrame.sort_values(by="列名", ascending=True)，by：根据哪一列排序，ascending=True：从小到大排列，ascending=False：从大到小排列
# 按照 duration_minutes 从小到大排列：
duration_ascending = clean_df.sort_values(by='duration_minutes', ascending=True).reset_index(drop=True)
print(f'按学习时长升序排列：\n{duration_ascending}')

# 按模块统计并排序投入时间，使用已经学过的 groupby() 统计各模块总时长
module_summary = clean_df.groupby('module')['duration_minutes'].sum().reset_index().rename(columns={'duration_minutes': 'total_minutes'})
print(f'各模块总学习时长：\n{module_summary}')

module_ascending = module_summary.sort_values(by='total_minutes', ascending=True)
print(f'各模块投入时间排名：\n{module_ascending}')

# 将各模块总分钟数转换为小时
module_ascending['total_hours'] = module_ascending['total_minutes'] / 60
module_ascending['total_hours'] = module_ascending['total_hours'].round(2)
print(f'各模块学习时长排名(小时)：\n{module_ascending}')

# 同时根据多个列排序，先按 week 升序排列，同一周内再按学习时长降序排列：
sort_df = clean_df.sort_values(by=['week', 'duration_minutes'], ascending=[True, False]).reset_index(drop=True)
print(f'按周数和学习时长排序：\n{sort_df}')
```
