# 练习1：Matplotlib 基础 + plt.figure()

```python
import matplotlib.pyplot as plt

# 创建一个宽 5 英寸、高 4.5 英寸的 Figure
plt.figure(figsize=(5, 4.5))

# 第一个列表 [1,2,3]：x 轴横坐标，三个点的 x 值依次是 (1、2、3)
# 第二个列表 [2,4,1]：y 轴纵坐标，三个点的 y 值依次是 (2、4、1)
plt.plot([1, 2, 3], [2, 4, 1])

plt.show()
```

# 练习2：plt.bar() 柱状图和plt.barh() 横向柱状图

```python
import matplotlib.pyplot as plt

weeks = ['week1', 'week2', 'week3']

hours = [12, 34, 23]

# plt.bar()柱状图
plt.figure(figsize=(5, 4.5))

plt.bar(weeks, hours)

# plt.barh()横向柱状图
plt.figure(figsize=(5, 4.5))

plt.barh(weeks, hours)

plt.show()
```

# 练习3：坐标轴名称与图表标题

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.figure(figsize=(5, 4.5))

plt.title('study hours by module')

plt.xlabel('study hours')

plt.ylabel('modules')

# 与第 2 周 pandas 连接起来
df = pd.DataFrame({
    "modules": ["Python", "Python", "NumPy", "NumPy", "Pandas", "Pandas"],
    "hours": [3, 5, 2, 4, 6, 3]})

moudule_hours = df.groupby('modules')['hours'].sum().reset_index()

plt.barh(moudule_hours['modules'], moudule_hours['hours'])

plt.show()
```

# 练习4：plt.plot() 折线图

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'weeks': ['week1', 'week2', 'week3', 'week4', 'week5'],
    'completion_rate': [60, 72, 68, 80, 92]
})

plt.figure(figsize=(5, 4.5))

plt.title('Weekly Task Completion Rate')

plt.xlabel('week')

plt.ylabel('completion_rate (%)')

plt.plot(df['weeks'], df['completion_rate'])

plt.show()
```

# 练习5：plt.hist() 直方图——数值都集中在哪些范围

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'module': [ "Python", "Python",
        "NumPy", "NumPy",
        "Pandas", "Pandas",
        "Matplotlib", "Matplotlib"],
    'number of users': [26, 34, 37, 44, 37, 55, 29, 49]
})

plt.figure(figsize=(5, 4.5))

plt.title('Number of users per module')

plt.xlabel('module')

plt.ylabel('number of users')

plt.hist(df['number of users'], bins=20) # bins 可以理解成：把数据范围分成多少个区间

plt.show()
```

# 练习6：plt.savefig()——把画好的图真正保存成 PNG 文件

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "modules": [
        "Python", "Python",
        "NumPy", "NumPy",
        "Pandas", "Pandas",
        "Matplotlib", "Matplotlib"
    ],
    "hours": [4, 5, 3, 4, 6, 3, 5, 4]
})

study_hours = df.groupby('modules')['hours'].sum().reset_index()

plt.figure(figsize=(5, 4.5))

plt.title('study_hours by modules')

plt.xlabel('study_hours')

plt.ylabel('modules')

plt.barh(study_hours['modules'], study_hours['hours'])
 
plt.tight_layout()  # 自动调整 Figure 内部元素之间的间距，尽量避免文字被挤压或截断

plt.savefig("outputs/figures/study_hours_by_modules.png", dpi=300)

plt.show()
```

# 练习7：

```python
import pandas as pd
import matplotlib.pyplot as plt

# 读取真实数据并确认字段
df = pd.read_excel('D:/Ai_Learning/python basic/data/processed/valid_study_log.xlsx')

print(df.columns)

# 统计每周总学习时长
week_hours = df.groupby('week')['duration_minutes'].sum().reset_index()

week_hours['week_label'] = 'Week' + week_hours['week'].astype(str)

week_hours['hours'] = week_hours['duration_minutes'] / 60

week_hours['hours'] = week_hours['hours'] .round(2)

plt.figure(figsize=(5, 4.5))

plt.title('Week_hours')

plt.xlabel('Week')

plt.ylabel('Hours')

plt.bar(week_hours['week_label'], week_hours['hours'])

plt.tight_layout()

plt.savefig("outputs/figures/week_hours.png", dpi=300)

plt.show()

# 不同模块总学习时长
module_hours = df.groupby('module')['duration_minutes'].sum().reset_index()

module_hours['hours'] = module_hours['duration_minutes'] / 60

module_hours['hours'] = module_hours['hours'].round(2)

plt.figure(figsize=(5, 4.5))

plt.title('Module_hours')

plt.xlabel('Hours')

plt.ylabel('Module')

plt.barh(module_hours['module'], module_hours['hours'])

plt.tight_layout()

plt.savefig("outputs/figures/module_hours.png", dpi=300)

plt.show()

# 计算每周任务完成率
completion_rate = df.groupby('week')['completed'].mean().reset_index()    # mean(): 完成任务占比

completion_rate['completion_rate'] = completion_rate['completed'] * 100

completion_rate['completion_rate'] = completion_rate['completion_rate'].round(2)

completion_rate['week_label'] = 'Week' + completion_rate['week'].astype(str) 

plt.figure(figsize=(5, 4.5))

plt.title('Completion_rate')

plt.xlabel('Week')

plt.ylabel('Completion_rate (%)')

plt.plot(completion_rate['week_label'], completion_rate['completion_rate'])

plt.tight_layout()

plt.savefig("outputs/figures/completion_rate.png", dpi=300)

plt.show()

# 学习时长分布
# 查看单次学习时长的基本统计信息
print(df['duration_minutes'].describe())

plt.figure(figsize=(5, 4.5))

plt.title('study_time_distribution')

plt.xlabel('Minutes')

plt.ylabel('Frequency')

plt.hist(df['duration_minutes'], bins=5)

plt.tight_layout()

plt.savefig("outputs/figures/study_time_distribution.png", dpi=300)

plt.show()
```
