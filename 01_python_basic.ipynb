## 练习4：for循环

# 练习6：函数

```python
# 说身高
students = {
    '小明': 1.95,
    '小花': 1.64,
    '小李': 1.76,
    '小娟': 1.43
}
def say_height(stu):
    if stu is None:
        print('目前没有同学测量身高')
    else:
        return sum(stu.values()) / len(stu)
for name, height in students.items():
    height_cm = height * 100
    print(f'大家好！我是{name}，我的身高是{height:.2f}米，也就是{height_cm:.0f}厘米')
average = say_height(students)
print(f'全班平均身高为：{average:.2f}米')


# 全班分数评级
people = {
    '小明': 95,
    '小花': 64,
    '小李': 76,
    '小娟': 43
}
def test_grades(grades):
    if grades is None:
        print('目前还未上传分数')
    else:
        if grades >= 85:
            return '优等'
        elif grades >= 60:
            return '中等'
        else:
            return '低等' 
for name, grade in people.items():
        level = test_grades(grade)
        print(f'我是{name}，我这次考了{grade}分，分数评级为{level}')


# 作业
## 加法运算
def add_tion(a, b):
    return a + b
print(f'2 + 3 = {add_tion(2, 3)}')
## 长度换算
def len_conversion(i):
    return i * 100
print(f'1米是{len_conversion(1)}厘米')
## 判断材料是否存在
materials = ["TiO2", "BiVO4", "ZnS"]
def check_material(materials, x):
    if x in materials:
        return f'存在材料{x}'
    return f'不存在材料{x}'
print(check_material(materials, 'TiO2'))
print(check_material(materials, 'SO2'))
# 计算平均电位
potentials = [-0.16, -0.23, -0.331, -0.282]
def cal_averagre(values):
    if not values:
        return None
    return sum(values) / len(values)
average = cal_averagre(potentials)
if average is None:
    print('没有点位数据')
else:
    print(f'平均电位为：{average:.2f}V')
```

# 练习7：txt文件的读写

```python
# 从 Python 标准库 pathlib 中导入 Path，Path 专门用于处理文件和文件夹路径
from pathlib import Path

# 设置 txt 文件的相对路径
file_path = Path('outputs/reports/note.txt')
# file_path.parent 表示 note.txt 所在的父文件夹，对应 outputs/reports
report_folder = file_path.parent
# 创建文件夹，parents=True：如果上层 outputs 也不存在，就一起创建，exist_ok=True：如果文件夹已经存在，不要报错
report_folder.mkdir(parents = True, exist_ok = True)
print(f'文件夹已经准备完毕：{report_folder}')

# 准备要写入 txt 文件的文字，三个引号可以保存多行文字
content = """第一周Python学习记录
已学习：
1. 变量
2. 列表
3. 字典
4. for 循环
5. if 判断
6. 函数
7. txt 文件读写
"""

# 把 content 中的文字写入 note.txt，encoding="utf-8" 用于保证中文正常显示，write_text 每次执行都会覆盖文件原来的内容
file_path.write_text(content, encoding = 'utf-8')
print(f'文件已经成功写入{file_path}')

# 从 txt 文件中读取全部文字，读取结果保存到变量 text 中
if file_path.exists(): # exists() 用于检查文件是否存在
    text = file_path.read_text(encoding = 'utf-8')
    print(text)
    print(type(text))
    
    # len(text) 统计字符串中的字符数量
    count = len(text)
    print(f'文本中有{count}个字符')

    # splitlines() 按照换行位置拆分文本
    lines = text.splitlines()
    
    # len(lines) 统计一共有多少行
    line_count = len(lines)
    print(f'文本一共有{line_count}行')
else:
    print(f'文件不存在，请先添加文件')

# 覆盖写入
file_path.write_text('这是覆盖写入文本', encoding = 'utf-8')
new_text = file_path.read_text(encoding='utf-8')
print(f'新的文本：{new_text}')

# 追加模式，读取模式，覆盖写入模式
# "w" 表示写入，文件已经存在时，会删除原有内容
with file_path.open('w', encoding='utf-8') as file:
    file.write('我是汪愈轩')
# "a" 表示追加，新内容会写在原文件末尾
with file_path.open('a', encoding='utf-8') as file:
    file.write('，今年24岁')
# "r" 表示只读取文件，read() 读取文件全部内容
with file_path.open('r', encoding='utf-8') as file:
    new_content = file.read()
print(new_content)
```

# 第一周综合题：材料阴极保护性能分析与报告生成

```python
# 现在有 5 种光电材料，需要根据其测试电位判断是否具有阴极保护趋势，并生成一份 txt 分析报告

# 304SS 的自腐蚀电位设为：
corrosion_potential = -0.16

# 样品测试数据如下：
samples = {
    "TiO2": [-0.12, -0.15, -0.14],
    "ZnIn2S4": [-0.28, -0.31, -0.30],
    "MgIn2S4": [-0.18, -0.21, -0.19],
    "BiVO4": [-0.10, -0.13, -0.11],
    "CdIn2S4": [-0.25, -0.27, -0.26]
}

# 任务 1：编写计算平均值的函数
def count_average(num):
    if not num:
        return None
    else:
        return sum(num) / len(num)

# 任务 2：编写性能判断函数
def performance_eva(num):
    if num is None:
        return None
    else:
        if num >= -0.16:
            return '性能较差'
        elif -0.16 > num > -0.25:
            return '性能一般'
        else:
            return '性能较好'

#对每个样品完成：输出样品名称；输出原始电位；输出平均电位；输出评价结果。
for material, potential in samples.items():

    # 输出样品名称
    print(f'样品名称：{material}')

    # 输出原始电位
    print(f'原始电位：{potential}V')

    # 输出平均电位
    average = count_average(potential)
    print(f'平均电位：{average:.2f}V')

    # 输出评价结果
    performance = performance_eva(average)
    print(f'评价结果：{performance}')
    print('=' * 35)

# 任务 4：找出平均电位最负的样品
best_material = None
best_average = None
for material, potential in samples.items():
    average = count_average(potential)
    if best_average is None or best_average > average:
        best_average = average
        best_material = material
print(f'目前保护效果最好的材料是{material}， 他的平均保护电位为{average:.2f}V')

# 任务 5：写入分析结果
from pathlib import Path
file_path = Path('outputs/reports/week_01_assignment.txt')
file_path.parent.mkdir(parents=True, exist_ok=True)

with file_path.open('w', encoding='utf-8') as file:

    file.write(f'材料阴极保护性能分析报告\n')

    for material, potential in samples.items():
        file.write('=' * 35 + '\n')

        file.write(f'样品名称：{material}\n')

        file.write(f'原始电位：{potential}V\n')

        average = count_average(potential)
        file.write(f'平均电位：{average:.2f}V\n')

        performance = performance_eva(average)
        file.write(f'评价结果：{performance}\n')

    file.write('=' * 35 + '\n')

    for material, potential in samples.items():
        average = count_average(potential)
        if best_average is None or best_average > average:
            best_average = average
            best_material = material        
    file.write((f'目前保护效果最好的材料是{material}， 他的平均保护电位为{average:.2f}V'))

# 任务 6：重新读取报告
if file_path.exists():

    report = file_path.read_text(encoding='utf-8')
    print(report)

    count_word = len(report)
    print(f'字符数：{count_word}')

    lines = report.splitlines()
    count_lines = len(lines)
    print(f'行数：{count_lines}')

else:
    print('文件不存在')
```
